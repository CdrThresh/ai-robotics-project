# Convert all point components to a DataFrame

```python
type_map = {
    1: "i1",
    2: "u1",
    3: "i2",
    4: "u2",
    5: "i4",
    6: "u4",
    7: "f4",
    8: "f8"
}

byte_order = ">" if msg.is_bigendian else "<"

dtype_fields = []

for f in msg.fields:
    base_type = np.dtype(byte_order + type_map[f.datatype])

    if f.count == 1:
        field_dtype = base_type
    else:
        field_dtype = (base_type, (f.count,))

    dtype_fields.append(
        (f.name, field_dtype, f.offset)
    )

point_dtype = np.dtype({
    "names": [x[0] for x in dtype_fields],
    "formats": [x[1] for x in dtype_fields],
    "offsets": [x[2] for x in dtype_fields],
    "itemsize": msg.point_step
})

points = np.frombuffer(
    bytes(msg.data),
    dtype=point_dtype,
    count=msg.width * msg.height
)

columns = {}

for name in points.dtype.names:
    values = points[name]

    if values.ndim == 1:
        columns[name] = values
    else:
        for i in range(values.shape[1]):
            columns[f"{name}_{i}"] = values[:, i]

df = pd.DataFrame(columns)

print("Number of points:", len(df))
print("Available columns:", df.columns.tolist())

display(df.head(20))
```

```python
import numpy as np
import pandas as pd
import open3d as o3d
import matplotlib.pyplot as plt

# Preserve original frame 0 data
df_features = df.copy(deep=True)

# Calculate distance from LiDAR sensor
df_features["range"] = np.sqrt(
    df_features["x"]**2 +
    df_features["y"]**2 +
    df_features["z"]**2
)

print("Range statistics (meters):")
print(df_features["range"].describe())

display(df_features.head(10))
```

# Calculate local height variation

```python
from scipy.spatial import cKDTree
import time

start = time.time()

# Extract original XYZ coordinates
xyz = df_features[["x", "y", "z"]].to_numpy(
    dtype=np.float64
)

# Only process finite XYZ coordinates
valid = np.isfinite(xyz).all(axis=1)
xyz_valid = xyz[valid]

# Build KD-tree
tree = cKDTree(xyz_valid)

# Number of neighboring points
k = 20

# Query neighbors
# Includes the point itself among its 20 nearest points
distances, indices = tree.query(
    xyz_valid,
    k=k,
    workers=-1
)

# Extract neighboring Z values
neighbor_z = xyz_valid[indices, 2]

# Calculate standard deviation
sigma_z = np.std(neighbor_z, axis=1)

# Add feature, preserving original row alignment
df_features["sigma_z"] = np.nan
df_features.loc[valid, "sigma_z"] = sigma_z

elapsed = time.time() - start

print(f"Processed {len(xyz_valid):,} points")
print(f"Processing time: {elapsed:.2f} seconds")

print("\nLocal height variation statistics:")
print(df_features["sigma_z"].describe())
```

# Prepare the five-feature dataset

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Five SOM features
feature_columns = [
    "x",
    "y",
    "z",
    "i",
    "sigma_z"
]

# Preserve the original data
df_som = df_features.dropna(
    subset=feature_columns
).copy()

# Remove infinite values
df_som = df_som[
    np.isfinite(df_som[feature_columns]).all(axis=1)
].copy()

# Extract features
X = df_som[feature_columns].to_numpy(dtype=np.float64)

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Group weights:
# 40% XYZ, 30% intensity, 30% height variation
feature_weights = np.array([
    0.40 / 3,  # X
    0.40 / 3,  # Y
    0.40 / 3,  # Z
    0.30,      # Intensity
    0.30       # sigma_z
])

# Apply weights for Euclidean-distance SOM training
X_weighted = X_scaled * np.sqrt(feature_weights)

# Use float32 to reduce memory consumption
X_weighted = X_weighted.astype(np.float32)

print("Number of points:", len(X_weighted))
print("Number of features:", X_weighted.shape[1])
print("Feature weights:", feature_weights)
print("SOM input shape:", X_weighted.shape)
```

# Configure and Train the SOM

```python
from minisom import MiniSom
import time

# SOM configuration
som_width = 20
som_height = 20
input_dim = X_weighted.shape[1]

# Reproducible training sample
rng = np.random.default_rng(42)

sample_size = min(15000, len(X_weighted))

sample_indices = rng.choice(
    len(X_weighted),
    size=sample_size,
    replace=False
)

X_train = X_weighted[sample_indices]

# Initialize SOM
som = MiniSom(
    x=som_width,
    y=som_height,
    input_len=input_dim,
    sigma=3.0,
    learning_rate=0.5,
    neighborhood_function="gaussian",
    random_seed=42
)

som.random_weights_init(X_train)

print("Training SOM...")
start = time.time()

som.train_random(
    X_train,
    num_iteration=10000,
    verbose=True
)

elapsed = time.time() - start

print(f"\nTraining completed in {elapsed:.2f} seconds")
print("SOM dimensions:", som_width, "x", som_height)
```

# Assign all LiDAR points to SOM neurons

```python
from sklearn.neighbors import NearestNeighbors

# Flatten SOM neuron weights
som_weights = som.get_weights().reshape(
    -1,
    input_dim
)

# Find nearest SOM neuron for each point
nn = NearestNeighbors(
    n_neighbors=1,
    algorithm="auto",
    n_jobs=-1
)

nn.fit(som_weights)

distances, bmu_indices = nn.kneighbors(X_weighted)

# Flatten results
bmu_indices = bmu_indices.ravel()
bmu_distances = distances.ravel()

# Convert flat indices to SOM grid coordinates
bmu_x = bmu_indices // som_height
bmu_y = bmu_indices % som_height

# Store assignments
df_som["som_neuron"] = bmu_indices
df_som["som_x"] = bmu_x
df_som["som_y"] = bmu_y
df_som["som_distance"] = bmu_distances

print("Assigned points:", len(df_som))
print("Unique SOM neurons used:", len(np.unique(bmu_indices)))
print("Mean BMU distance:", bmu_distances.mean())

display(df_som.head())
```

# CLutering into 10 SOM groups

```python
import numpy as np
from sklearn.cluster import KMeans

# Extract the trained SOM neuron weights
som_weights = som.get_weights().reshape(
    -1, X_weighted.shape[1]
)

# Cluster the SOM neurons
n_clusters = 10

kmeans = KMeans(
    n_clusters=n_clusters,
    random_state=42,
    n_init=10
)

neuron_cluster_labels = kmeans.fit_predict(som_weights)

# Map each LiDAR point's BMU to a cluster
som_cluster_labels = neuron_cluster_labels[bmu_indices]

print("Number of SOM groups:", n_clusters)
print("Number of LiDAR points:", len(som_cluster_labels))

for cluster_id in range(n_clusters):
    count = np.sum(som_cluster_labels == cluster_id)
    print(f"Group {cluster_id}: {count:,} points")
```

```python
feature_columns = ["x", "y", "z", "i", "sigma_z"]

df_som_20m = df_20m.dropna(
    subset=feature_columns
).copy()

X_20m = df_som_20m[feature_columns].to_numpy(
    dtype=np.float64
)

# Normalize
scaler_20m = StandardScaler()
X_scaled_20m = scaler_20m.fit_transform(X_20m)

# Same weights as original experiment
feature_weights = np.array([
    0.40 / 3,
    0.40 / 3,
    0.40 / 3,
    0.30,
    0.30
])

X_weighted_20m = (
    X_scaled_20m * np.sqrt(feature_weights)
).astype(np.float32)

print("SOM input shape:", X_weighted_20m.shape)
```

# 3 Train the 20 x 20 SOM

```python
from minisom import MiniSom
import time

som_20m = MiniSom(
    x=20,
    y=20,
    input_len=5,
    sigma=3.0,
    learning_rate=0.5,
    neighborhood_function="gaussian",
    random_seed=42
)

rng = np.random.default_rng(42)

sample_indices = rng.choice(
    len(X_weighted_20m),
    size=min(15000, len(X_weighted_20m)),
    replace=False
)

X_train_20m = X_weighted_20m[sample_indices]

som_20m.random_weights_init(X_train_20m)

start = time.time()

som_20m.train_random(
    X_train_20m,
    num_iteration=10000,
    verbose=True
)

print(f"Training time: {time.time() - start:.2f} seconds")
```

# 4 Assign points and group neurons into 10 clusters

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.cluster import KMeans

# Extract neuron weights
neuron_weights = som_20m.get_weights().reshape(-1, 5)

# Assign each LiDAR point to its nearest neuron
nn = NearestNeighbors(n_neighbors=1)
nn.fit(neuron_weights)

_, indices = nn.kneighbors(X_weighted_20m)
bmu_20m = indices.ravel()

# Group SOM neurons into 10 clusters
kmeans_20m = KMeans(
    n_clusters=10,
    random_state=42,
    n_init=10
)

neuron_groups_20m = kmeans_20m.fit_predict(
    neuron_weights
)

# Map groups back to original LiDAR points
som_labels_20m = neuron_groups_20m[bmu_20m]

print("Points assigned:", len(som_labels_20m))
print("Unique SOM groups:", len(np.unique(som_labels_20m)))
```

```python
import numpy as np
import open3d as o3d

# Original 20-meter point cloud
xyz = df_som_20m[["x", "y", "z"]].to_numpy(dtype=np.float64)

# Create point cloud
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(xyz.copy())

# RANSAC ground plane detection
plane_model, ground_indices = pcd.segment_plane(
    distance_threshold=0.15,
    ransac_n=3,
    num_iterations=1000
)

a, b, c, d = plane_model
normal = np.array([a, b, c])
normal = normal / np.linalg.norm(normal)

print("Ground plane normal:", normal)

# Check that the detected plane is approximately horizontal
if abs(normal[2]) < 0.85:
    raise ValueError(
        "Detected plane may not be ground. Inspect before continuing."
    )

# Preserve original indices
non_ground_mask = np.ones(len(xyz), dtype=bool)
non_ground_mask[ground_indices] = False

non_ground_indices = np.flatnonzero(non_ground_mask)

xyz_non_ground = xyz[non_ground_indices]

print(f"Original points: {len(xyz):,}")
print(f"Ground points removed: {len(ground_indices):,}")
print(f"Remaining points: {len(xyz_non_ground):,}")
```
